# PCA (Principal Component Analysis)


```python
from pathlib import Path
import re
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

# Return dictionary with all words of a file and their count
def get_word_count(file):
    dict = {}
    for word in file.read_text().split():
        parsed_word = re.sub(r"[^\w\s]", "", word.lower())
        if parsed_word: dict[parsed_word] = dict.get(parsed_word, 0) + 1
    return sorted(dict.items(), key=lambda x: x[1], reverse=True)

# count words for each text file
combined = {}
for file in Path("../assets/textfiles").glob("*.txt"):
    # use regex to parse author and title
    author = re.search(r"(.+) -", file.name).group(1)
    title = re.search(r"- (.+)\.", file.name).group(1)

    dict = get_word_count(file)
    print(f"  - {author}->{title}: {dict}")

    # combine all word counts
    for word, count in dict:
        combined[word] = count
        combined[word] = combined.get(word, 0) + count

# sort and print combined word count
combined = sorted(combined.items(), key=lambda x: x[1], reverse=True)

# print top 50 words
print(f"Top 50 words: {combined[:50]}")
```
